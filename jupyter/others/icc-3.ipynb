```python
"""
Python - OOP
"""
class dog:

  # 添加类属性，所有实例共享
  para = "shared in all instances"

  # 添加实例属性，每个实例独有
  def __init__(self,voice,action):
    """
    初始化方法，构造函数
    初始化dog对象
    当创建dog实例时自动调用
    """
    self.voice = voice
    self.action = action
  
  # 添加实例方法
  def bark(self):
    return f"{self.voice} says woof,"
  def getinfo(self):
    return f"{self.voice} is {self.action}"

mydog = dog("bark", "b")
print(mydog.voice, mydog.action, mydog.para)
print(mydog.bark(), mydog.getinfo())
# print(mydog)
```

```python
"""Inheritance
"""
# 父类（基类/超类）
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return "Some sound"
    
    def info(self):
        return f"I am {self.name}"

# 子类（派生类）
class Dog(Animal):  # 继承 Animal
    def speak(self):
        """重写（Override）父类方法"""
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

# 使用
dog = Dog("Buddy")
cat = Cat("Whiskers")

print(dog.speak())  # "Woof!"（子类方法）
print(dog.info())   # "I am Buddy"（继承的方法）
print(cat.speak())  # "Meow!"

# 类型检查
print(isinstance(dog, Dog))     # True
print(isinstance(dog, Animal))  # True（子类是父类的实例）
print(isinstance(dog, Cat))     # False

print(issubclass(Dog, Animal))  # True
print(issubclass(Dog, Cat))     # False
```
