```python
# [流浪地球](https://blog.csdn.net/banxia_frontend/article/details/151763512)
from collections import deque

N, E = map(int, input().split())

manual = [] # 手动启动发动机列表
for _ in range(E):
    T, P = map(int, input().split())
    manual.append((T, P))

# 1. 按时间升序排序
manual.sort(key=lambda x: x[0])

# 2. 初始化每个发动机的启动时间数组，inf表示未启动
time = [float('inf')] * N

# 3. 初始化启动队列，内部包含启动的发动机
q = deque()
for t, p in manual:
    if time[p] > t:  # 如果手动启动的发动机启动时间大于
        time[p] = t
        q.append((p, t))

# 4. BFS 扩散
while q: # 有未开始感染的感染源需要处理
    cur, t = q.popleft()
    for nxt in [(cur - 1 + N) % N, (cur + 1) % N]: # 在左右开始感染
        if time[nxt] > t + 1: 
            time[nxt] = t + 1 # 更新感染时间
            q.append((nxt, t + 1)) # 加入感染源，去感染下一个未启动的发动机

# 5. 找出最后被启动的发动机
max_t = max(time)
res = [i for i, v in enumerate(time) if v == max_t]

print(len(res))
print(" ".join(map(str, res)))

# 输入
# 8 2
# 0 2
# 0 6
```

```python
# 流量波峰 https://blog.csdn.net/banxia_frontend/article/details/151763394
# 输入： 3 5 4 7 2 1
nums = list(map(int, input().split()))
n = len(nums)
res = [float('inf')]*n
for i, e in enumerate(nums):
  l,r = i-1,i+1
  if l<0 or r<0 or l>=n or r>=n:
    continue 
  elif nums[l]>=nums[i] or nums[r]>=nums[i]:
    continue 
  while nums[l]<nums[i] and nums[l-1]<nums[l] and l>=1:
    l-=1
  while nums[r]>nums[i] and nums[r]>nums[r+1] and r<n-1:
    r+=1
  res[i] = r-l
# print(res)
print(min(res)) if not  min(res) == float('inf') else print(-1)
```

```python
# 增强的strstr https://blog.csdn.net/banxia_frontend/article/details/151763731
# 输入
# abcd
# b[cd]
s = input()
nums = input()
l, r = nums.index('['), nums.index(']')
res = [] 
left, right = nums[:l], nums[r+1:]
for e in nums[l+1:r]:
  res.append(left+e+right)
print(res)
arr = []
for e in res:
  if e in s:
    arr.append(s.index(e))
print(min(arr))
```
